```python

df['Day'] = df['Day'].fillna(0)
df['Order_Quantity'] = df['Order_Quantity'].fillna(df['Order_Quantity'].mean())
```

```python
df['Product_Description'] = df['Product_Description'].fillna('Unknown')
df['Age_Group'] = df['Age_Group'].fillna(0)
```

```python
df.isnull().sum()
```

```python
df['Country'].unique()
```

```python
df[df.duplicated()]
df
```

```python
df['Model'] = df['Product_Description'].str.split(' ').str[0]
df['Colour'] = df['Product_Description'].str.split(' ').str[1]
df['Size'] = df['Product_Description'].str.split(',').str[1]
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Colour'] = df['Colour'].fillna(0)
df['Size'] = df['Size'].fillna(0)
```

```python
df['Model'] = df['Model'].astype(str)
df['Colour'] = df['Colour'].astype(str)
df['Size'] = df['Size'].astype(int)
```

```python
df.info()
```

```python
df.drop('Product_Description', axis = 1, inplace = True)
```

```python
df.head()
```

```python
df.rename(columns = {'Size' : 'Serial'}, inplace = True)
df
```

```python
df['Age_Group'].unique()
```

```python
df['Age_Group'].mode
```

```python
df.columns = df.columns.str.strip()
```

```python
df.columns.unique()
```

```python
df['Profit'] = df['Profit'].str.replace('$', '').str.replace(',', '')
df['Revenue'] = df['Revenue'].str.replace('$', '').str.replace(',', '')
df['Unit_Cost'] = df['Unit_Cost'].str.replace('$', '').str.replace(',', '')
df['Unit_Price'] = df['Unit_Price'].str.replace('$', '').str.replace(',', '')
df['Cost'] = df['Cost'].str.replace('$', '').str.replace(',', '')
```

```python
df['Profit'] = df['Profit'].astype(float)
df['Revenue'] = df['Revenue'].astype(float)
df['Unit_Cost'] = df['Unit_Cost'].astype(float)
df['Unit_Price'] = df['Unit_Price'].astype(float)
df['Cost'] = df['Cost'].astype(float)
```

```python
df['Profit_Margin'] = (df['Profit'] / df['Revenue']) * 100
```
